# CRYOWRF validation against the AWS

Surface meteorology at PEA: time series, scatter plots and summary statistics.
Wind speed from the CRYOWRF fine mesh at anemometer height (`VW_SFC`), wind direction from the
WRF 10 m diagnostic (`wd10_deg`), other variables from the WRF 2 m / surface diagnostics.
Metrics: MAE, bias, r, PBIAS % = 100·Σ(mod−obs)/Σ(obs), NRMSE % = 100·RMSE/mean(obs)
(no percentage for wind direction; temperature in K for the percentages).
All functions are in `dbs_utils.py`.

```python
# ── Run switches ─────────────────────────────────────────────────────────────
SIM_KEY    = "trans_syn"   # event used when RUN_ALL is False
RUN_ALL    = True          # True -> run every event; False -> only SIM_KEY
SAVE_PLOTS = True          # True -> write figures to OUTPUT_DIR
DOMAIN     = "d05"

import importlib
import dbs_utils as ru
importlib.reload(ru)

RUN_KEYS = ru.get_run_keys(SIM_KEY, RUN_ALL)
print("Events to run:", RUN_KEYS)
print("Save figures :", SAVE_PLOTS)
```

## 1. Build data

```python
merged_by_event = {}
for key in RUN_KEYS:
    print(f"Building {key} ...")
    merged_by_event[key] = ru.build_merged_event(key, domain=DOMAIN)
    print(f"  {len(merged_by_event[key])} timesteps")
```

## 2. Per-event time series and scatter

```python
for key in RUN_KEYS:
    ru.plot_timeseries(merged_by_event[key], domain=DOMAIN, save=SAVE_PLOTS)
    ru.plot_scatter_event(merged_by_event[key], domain=DOMAIN, save=SAVE_PLOTS)
```

## 3. Combined scatter

```python
# Figure style for the combined scatter
style = dict(
    fig_w        = 30,     # figure width (in)
    fig_h        = 6.8,    # figure height (in)
    suptitle_fs  = 22,     # figure title
    title_fs     = 18,     # panel titles
    label_fs     = 16,     # axis labels
    tick_fs      = 14,     # tick labels
    legend_fs    = 16,     # legend
    r_fs         = 17,     # "r = ..." annotation
    point_size   = None,   # marker size (None = default)
    point_alpha  = None,   # marker alpha (None = default)
    one_to_one_lw= 1.5,    # 1:1 line width
    title_pad    = 8,      # title padding
    circular_wd  = True,   # circular r for wind direction (as in the table)
)

ru.plot_scatter_combined(merged_by_event, domain=DOMAIN, save=SAVE_PLOTS, **style)
```
